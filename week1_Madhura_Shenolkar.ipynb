### 4.3 Eigenvalues & Eigenvectors

For the matrix:

1. Compute eigenvalues and eigenvectors using `np.linalg.eig`.

4. **Explain** in a markdown cell (below): what does it geometrically mean for a matrix to stretch a vector?

````python
import matplotlib.pyplot as plt
A = np.array([[4, 1], [2, 3]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)  # YOUR CODE HERE

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors (columns):\n{eigenvectors}")

# Verify Av = λv for each eigenpair
# YOUR CODE HERE
for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]

    Av = A @ v
    lv = lam * v

    print(f"\nEigenpair {i+1}:")
    print("A @ v =", Av)
    print("λ * v =", lv)
    assert np.allclose(Av, lv), "Av != λv"
# Plot
#fig, ax = plt.subplots()

for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    scaled_v = lam * v

    ax.quiver(0, 0,
              scaled_v[0], scaled_v[1],
              angles='xy', scale_units='xy', scale=1,
              label=f'λ={lam:.2f}')

ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.axhline(0, color='black', lw=0.8)
ax.axvline(0, color='black', lw=0.8)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Eigenvectors Scaled by Eigenvalues')
ax.legend()
plt.grid()
plt.show()
print("4.3 passed ✓")
````

**Your Geometric Explanation (4.3):**

> A matrix transformation can rotate, stretch, compress, or shear vectors in space. Most vectors change both their direction and their length when a matrix is applied to them. However, some special vectors, called **eigenvectors**, keep their direction unchanged. They are only scaled (stretched or squished) by a certain amount.

The amount by which an eigenvector is scaled is called its **eigenvalue**.

- If the eigenvalue is greater than 1, the matrix stretches the eigenvector.
- If the eigenvalue is between 0 and 1, the matrix compresses (squishes) the eigenvector.
- If the eigenvalue is negative, the vector is reversed in direction and scaled by the magnitude of the eigenvalue.
- If the eigenvalue is 1, the vector remains unchanged.

Geometrically, eigenvectors represent the directions that are preserved by the transformation, while eigenvalues tell us how much those directions are stretched or compressed. Thus, a matrix "stretches" or "squishes" space differently along its eigenvector directions.

### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.
3. **Reconstruct** `X` from the SVD and verify it matches the original.
4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(X, full_matrices=False) # YOUR CODE HERE

X_reconstructed = U @ np.diag(S) @ Vt # YOUR CODE HERE — reconstruct from U, S, Vt
X_approx        = S[0] * np.outer(U[:, 0], Vt[0, :])  # YOUR CODE HERE — rank-1 approximation

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{X_approx}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

**SVD → PCA Connection (4.4):**

> *Explain: what are U, S, Vt in SVD, and how do they relate to principal components? Which matrix contains the directions of maximum variance?*

Singular Value Decomposition (SVD) factorizes a data matrix X into

X = UΣVᵀ

where:

- **U** contains the left singular vectors. These represent the directions of the transformed samples (coordinates in the new space).
- **S (Σ)** contains the singular values, which indicate how much variance or information is captured along each direction.
- **Vᵀ** contains the right singular vectors. Its rows (equivalently, the columns of V) represent the principal directions in feature space.

PCA seeks directions that capture the maximum variance in the data. When SVD is applied to centered data:

- The **columns of V** (or rows of Vᵀ) are the **principal components**.
- The **singular values** determine how much variance is explained by each principal component (variance ∝ singular value²).
- The matrix **UΣ** gives the coordinates of the observations projected onto those principal components.

Therefore, the matrix **V (or Vᵀ)** contains the directions of maximum variance. Keeping only the first few columns of V (those corresponding to the largest singular values) performs dimensionality reduction while preserving most of the information in the data.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = stats.kstest(s1, 'norm')  # YOUR CODE HERE
ks_s2 = stats.kstest(s2, 'norm')  # YOUR CODE HERE
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test on non-stationary series
t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = adfuller(ts) # YOUR CODE HERE — adfuller(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff = np.diff(ts) # YOUR CODE HERE — np.diff
adf_diff = adfuller(ts_diff)  # YOUR CODE HERE
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

````python
from scipy import stats
from statsmodels.tsa.stattools import adfuller
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    # YOUR CODE HERE
    epsilon = 1e-10

    bin_edges = np.histogram_bin_edges(expected, bins=bins)

    expected_counts, _ = np.histogram(expected, bins=bin_edges)
    actual_counts, _ = np.histogram(actual, bins=bin_edges)


    expected_pct = expected_counts / len(expected)
    actual_pct = actual_counts / len(actual)

    expected_pct += epsilon
    actual_pct += epsilon

    psi = np.sum((actual_pct - expected_pct) *
                 np.log(actual_pct / expected_pct))

    return psi
    pass

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)    # training distribution
drift_dist = np.random.normal(65, 12, 1000)    # shifted production distribution

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
fig, ax = plt.subplots()
# YOUR CODE HERE
ax.hist(train_dist, bins=30, density=True, alpha=0.5, label='Training Distribution')
ax.hist(drift_dist, bins=30, density=True, alpha=0.5, label='Production Distribution')
ax.legend(); ax.set_title('Distribution Shift Visualization')
plt.tight_layout(); plt.show()
````

**Concept Drift vs Covariate Drift & PSI Thresholds (5.5):**

- **Concept drift:** A change in the relationship between input features and the target variable over time, causing the model's predictions to become inaccurate even if the input distributions remain the same.
- **Covariate drift:**  A change in the distribution of the input features (X) while the relationship between features and the target remains unchanged.

- **PSI < 0.1:** Stable distribution — no significant drift detected; retraining is usually unnecessary.
- **PSI 0.1–0.2:** Moderate (minor) drift — monitor model performance closely and consider retraining if accuracy degrades.
- **PSI > 0.2:** Major drift — significant distribution shift; model retraining or recalibration is recommended.

- **Retraining trigger example:** If the PSI for customer age or transaction amount exceeds 0.2 for several consecutive weeks and the model's accuracy drops below a predefined threshold, retrain the model using the latest production data.

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability
p_free = p_free_given_spam * p_spam + p_free_given_ham * p_ham # YOUR CODE HERE

# Bayes
p_spam_given_free = (p_free_given_spam * p_spam) / p_free  # YOUR CODE HERE

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    # YOUR CODE HERE
    prior_ham = 1 - prior_spam

    p_word = p_word_given_spam * prior_spam + \
             p_word_given_ham * prior_ham


    posterior = (p_word_given_spam * prior_spam) / p_word

    return posterior
    pass

pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

### Bayes' Theorem in the Spam Filter Example

- **Prior:** P(Spam) = 0.30
  - Before seeing any words, 30% of emails are spam.

- **Likelihood:** P("free" | Spam) = 0.80
  - Spam emails contain the word "free" 80% of the time.

- **Evidence:** P("free") = 0.275
  - The overall probability that an email contains the word "free".

- **Posterior:** P(Spam | "free") = 0.8727
  - After observing the word "free", the probability that the email is spam increases to about 87.3%.

Thus, Bayes' theorem updates our belief about whether an email is spam based on the evidence provided by the word "free".

### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples = 5000
sample_size = 30
sample_means = np.array([
    np.mean(np.random.choice(population, size=sample_size))
    for _ in range(n_samples)
])# YOUR CODE HERE — array of 5000 sample means

pop_mean = population.mean()
pop_std  = population.std()
clt_std  = pop_std / np.sqrt(sample_size)# YOUR CODE HERE — predicted std of sample means

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots()
# YOUR CODE HERE — histogram of sample means + theoretical normal overlay
ax.hist(sample_means, bins=40, density=True, alpha=0.6, label='Sample Means')
x = np.linspace(sample_means.min(), sample_means.max(), 300)
ax.plot(x, stats.norm.pdf(x, loc=pop_mean, scale=clt_std),
        linewidth=2, label='CLT Normal Approximation')
ax.set_title('CLT: Distribution of Sample Means')
plt.tight_layout(); plt.show()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

**CLT Reflection (6.4):**

> *In 2–3 sentences: Why does this result matter for statistics and machine learning? What assumption does the CLT justify in many classical statistical tests?*

The Central Limit Theorem is important because it tells us that the distribution of sample means becomes approximately normal, even when the original population is not normally distributed. This allows us to use normal-based methods for estimation and hypothesis testing in statistics and machine learning. The CLT justifies the assumption of approximate normality underlying many classical statistical procedures, such as confidence intervals, t-tests, ANOVA, and linear regression inference.
